## 3.loc与iloc

```python

dates = pd.date_range('2024-01-02', periods=10, freq='B')  # 生成10个工作日的日期范围
df = pd.DataFrame(
    np.random.normal(0, 0.02, size=(10, 3)),
    index=dates,
    columns=['000001', '000002', '600519']
)
print(df)
```

```python
# loc: 按标签选择数据
print(df.loc['2024-01-02'])  # 选择2024-01-02的行
print(df.loc['2024-01-02', '000001'])  # 选择2024-01-02的000001列的值
print(df.loc['2024-01-02':'2024-01-04'])  # 选择2024-01-02到2024-01-04的行(包含2024-01-04)
print(df.loc['2024-01-03':'2024-01-05', ['000001', '600519']])  
print(df.loc[:, ['000001', '600519']])  # 选择所有行的000001和600519列
print(df.loc[df['000001'] > 0])  # 选择000001列大于0的行
```

```python
# iloc: 按位置选择数据
print(df.iloc[0])  # 选择第一行
print(df.iloc[0, 0])  # 选择第一行第一列的值
print(df.iloc[0:3])  # 选择前三行
print(df.iloc[0:3, 0:2])  # 选择前三行前两列的值
print(df.iloc[[0, 2, 4], [0, 2]])  # 选择第一行和第三行的第一列和第三列的值
```

```python
# at/iat: 按标签/位置选择单个值
print(df.at['2024-01-02', '000001'])  # 按标签选择单个值
print(df.iat[0, 0])  # 按位置选择单个值
```

## 4.布尔索引与 query

```python
print(df[df['000001'] > 0])  # 选择000001列大于0的行

mask = (df['000001'] > 0) & (df['000002'] < 0) # 选择000001列大于0且000002列小于0的行的布尔索引
print(mask)  # 选择000001列大于0且000002列小于0的行的布尔索引
print(df[mask])  # 选择000001列大于0且000002列小于0的行

print(df[['000001', '600519']])  # 选择000001和600519两列

print(df[df['000001'].between(-0.01, 0.01)])  # 选择000001列在-0.01和0.01之间的行

print(df.query('`000001` > 0 and `000002` < 0'))  # 使用query方法选择000001列大于0且000002列小于0的行
```

## 5. 缺失值、重复值、类型转换（数据清洗）

```python
# 构造带缺失值的 DataFrame
df = pd.DataFrame({
    'date': pd.to_datetime(['2024-01-02', '2024-01-02', '2024-01-03', '2024-01-03', '2024-01-03']),
    'code': ['000001', '000001', '000001', '000002', '000002'],
    'close': [10.0, 10.0, np.nan, 20.0, 20.5],
    'volume': [1000, 1000, 1100, np.nan, 2100],
})
print(df)

# 检查缺失值
print(df.isna())  # 检查缺失值，返回布尔值 
print(df.isna().sum())  # 检查缺失值，返回每列缺失值的数量

# 删除缺失值
print(df.dropna())  # 删除包含缺失值的行
print(df.dropna(how='all'))  # 删除所有值都为缺失值的行
print(df.dropna(subset=['close']))  # 删除close列中包含缺失值的行

# 填充缺失值
print(df.fillna(0))  # 用0填充缺失值
print(df.ffill())  # 用前一个有效值填充缺失值
print(df.bfill())  # 用后一个有效值填充缺失值
print(df.groupby('code')['close'].ffill())  # 按code分组，用前一个有效值填充缺失值
```

```python
# 删除重复值
mask = df.duplicated(subset=['date', 'code'])  #同一支股票在同一天可能有多条记录，只保留一条
print(mask)  # 检查重复值，返回布尔值
print(df[mask])  # 显示重复值
df_clean = df.drop_duplicates(subset=['date', 'code'])  # 删除重复值
print(df_clean)  # 显示删除重复值后的DataFrame

# 类型转换
print(df.dtypes)  # 查看数据类型
df['code'] = df['code'].astype(str)  # 将code列转换为字符串类型
```

以下是七步走进行数据清洗的代码：

```python
df = df_raw.copy()  # 复制一份数据，避免修改原始数据

# Step1: 日期解析，解析失败变成NaT并删除
df['date'] = pd.to_datetime(df['date'],errors='coerce')  # 解析日期，解析失败变成NaT
df = df.dropna(subset=['date'])  # 删除date列中为NaT的行
print("Step1: 日期解析后形状：", df.shape)

# Step2: 股票代码转字符串，补零至6位
df['code'] = df['code'].astype(str).str.zfill(6) # 转字符串并补零至6位
print("Step2: 股票代码转字符串并补零后形状：", df.shape)

# Step3: close列和volume列转浮点数，无法转的变成NaN
df['close'] = pd.to_numeric(df['close'], errors='coerce')
df['volume'] = pd.to_numeric(df['volume'], errors='coerce')
print("Step3: close列和volume列转浮点数后形状：", df.shape)
print("close列缺失值数量：", df['close'].isna().sum())
print("volume列缺失值数量：", df['volume'].isna().sum())

# Step4: 删除close列小于等于0的行
df = df.drop(df[df['close'] <= 0].index)
print("Step4: 删除close列小于等于0的行后形状：", df.shape)

# Step5: 删除重复行，保留第一次出现的
df = df.drop_duplicates(subset=['date', 'code'], keep='first')
print("Step5: 删除重复行后形状：", df.shape)

# Step6: 按code和date排序
df = df.sort_values(['code', 'date']).reset_index(drop=True)
print("Step6: 按code和date排序后形状：", df.shape)

# Step7: 按股票分组做缺失值填充
df['close'] = df.groupby('code')['close'].ffill()  # 用前一个有效值填充缺失值
df['volume'] = df.groupby('code')['volume'].ffill()  # 用前一个有效值填充缺失值
print("Step7: 按股票分组做缺失值填充后形状：", df.shape)
print("close列缺失值数量：", df['close'].isna().sum())
print("volume列缺失值数量：", df['volume'].isna().sum())

print("清洗后的数据预览：")
print(df)
```

## 6. groupby聚合

### 6.1 基础聚合

```python
df = pd.DataFrame({
    'date': pd.to_datetime(['2024-01-02', '2024-01-02', '2024-01-03', '2024-01-03', '2024-01-04', '2024-01-04']),
    'code': ['000001', '000002', '000001', '000002', '000001', '000002'],
    'ret': [0.01, -0.02, 0.03, 0.01, -0.01, 0.02],
    'volume': [1000, 2000, 1100, 2100, 1200, 2200],
})
print(df)

# 按股票分组求收益均值(单列单聚合)
print(df.groupby('code')['ret'].mean())

# 多列单聚合
print(df.groupby('code')[['ret', 'volume']].mean())

# 多列多聚合
print(df.groupby('code').agg({
    'ret': ['mean', 'std', 'max'],
    'volume': ['mean', 'max'],
})
)

# 自定义列名
print(df.groupby('code').agg(
    ret_mean = ('ret', 'mean'),
    ret_std = ('ret', 'std'),
    ret_max = ('ret', 'max'),
    vol_mean = ('volume', 'mean'),
    vol_max = ('volume', 'max'),
)
)
```

### 6.2 transform与apply

```python
# transform函数作用是对DataFrame的每一列进行操作，返回一个新的DataFrame
df['ret_demeaned'] = df.groupby('code')['ret'].transform(lambda x: x - x.mean()) 
    # 计算每个code的ret的均值，并将每个code的ret减去均值
    # .transform()函数对每一列进行操作，返回一个新的DataFrame
    # lambda x：x - x.mean() 可以看作一个函数，输入值x是一个Series（这里面的元素依次传给x），输出值是x - x.mean()
print(df)
print(df['ret_demeaned'])
# 另一种写法
df['ret_demeaned2'] = df['ret'] - df.groupby('code')['ret'].transform('mean') 
print(df)# 计算每个code的ret的均值，并将每个code的ret减去均值

# apply和transform
series_mean = df.groupby('code')['ret'].apply(lambda x: x - x.mean()) 
    # .apply()函数对每一列进行操作，返回一个Series
print(series_mean)# 计算每个code的ret的均值，并将每个code的ret减去均值


```

在量化中，transform常用于对数据标准化。下面是一个例子：

```python
# 计算每只股票的zscore
df['ret_z'] = df.groupby('code')['ret'].transform(lambda x: (x - x.mean()) / x.std())
print(df)
```

### 8.3 shift 与 pct_change

```python
# shift：把序列整体平移
prices = pd.Series([100, 101, 102, 103, 104])
print(prices.shift(1))  # 向后平移一位
print(prices.shift(-1)) # 向前平移一位

# pct_change：计算百分比变化
print(prices.pct_change())  # 简单收益率
print(np.log(prices / prices.shift(1)))  # 对数收益率
print(np.log(prices).diff()) # 对数收益率的另一种写法
```

### 8.4 rolling：滚动窗口

```python
idx = pd.date_range('2024-01-01',periods=100, freq='B')  # 生成工作日日期范围
ret = pd.Series(np.random.normal(0, 0.02, size=100), index=idx, name='ret  ')  # 生成随机数序列

# 20日滚动均值与标准差
rolling_mean = ret.rolling(window=10).mean()  #前九行由于不满10个数据点，滚动均值为NaN
rolling_std = ret.rolling(window=10).std()
rolling_std2 = ret.rolling(window=10, min_periods=1).std()  #注意区别
print(rolling_mean.head(20))
print(rolling_std.head(20))
print(rolling_std2.head(20))

# 两个Series的滚动相关系数
ret2 = pd.Series(np.random.normal(0, 0.02, size=100), index=idx, name='ret2')
rolling_corr = ret.rolling(window=10).corr(ret2)
print(rolling_corr.tail())
```

### 8.5 expanding：扩展窗口

```python
expanding_mean = ret.expanding().mean()  # 每次计算均值的window不断扩大
print(expanding_mean.tail())
```

### 8.6 ewm：指数加权移动平均

```python
# ewm：从起点到当前，**越近的数据权重越高，指数衰减**，不需要固定窗口。
#公式：ewm_mean[t] = (1 - alpha) * ewm_mean[t-1] + alpha * ret[t]
ewm_mean = ret.ewm(span=20, adjust=False).mean()  
    # 指数加权均值，span是衰减因子，adjust=False表示不调整权重
    # 这里规定，alpha=2/(span+1)，目的是让span较大时，(1-alpha)**span趋近0.5
ewm_vol = ret.ewm(span=20, adjust=False).std()  
print(ewm_mean.tail())
print(ewm_vol.tail())
```
