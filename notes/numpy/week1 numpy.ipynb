## 2.索引、切片、布尔索引

```python
import numpy as np

ret = np.array([
    [0.01, 0.02, -0.01, 0.00],
    [0.02, -0.01, 0.03, 0.01],
    [-0.01, 0.00, 0.01, 0.02],
])

# 1. 单个元素
print(ret[0, 1])       # 第 0 天，第 1 只股票

# 2. 取一行（某一天所有股票）
print(ret[0, :])

# 3. 取一列（某只股票所有天）
print(ret[:, 1])

# 4. 切片：前 2 天、前 3 只股票
print(ret[:2, :3])

# 5. 布尔索引：找出所有大于 0 的收益率
mask = ret > 0
print(mask)
print(ret[mask])

# 6. 按条件替换：把负收益设为 0
ret_clipped = ret.copy()
ret_clipped[ret_clipped < 0] = 0
print(ret_clipped)
```

```python
import numpy as np

ret = np.array([
    [0.01, 0.02, -0.01, 0.00],
    [0.02, -0.01, 0.03, 0.01],
    [-0.01, 0.00, 0.01, 0.02],
])

# 找出某只股票收益率为正的交易日
stock_ret = ret[:, 1]
positive_days = stock_ret[stock_ret > 0]
print(positive_days)

# 找出极端收益
extreme = ret[np.abs(ret) > 0.015]
print(extreme)
```

## 3.轴 axis 与聚合

```python
import numpy as np

ret= np.array([
    [0.01, 0.02, -0.01, 0.00],
    [0.02, -0.01, 0.03, 0.01],
    [-0.01, 0.00, 0.01, 0.02],
])

print(ret.mean())          # 所有元素的均值
print(ret.mean(axis=0))    # 按列求均值（每只股票的平均收益率）
print(ret.mean(axis=1))    # 按行求均值（每一天的平均收益率）   

print(ret.std())           # 所有元素的标准差
print(ret.std(axis=0))     # 按列求标准差（每只股票的收益率波动）   
print(ret.std(axis=1))     # 按行求标准差（每一天的收益率波动）

print(ret.sum())           # 所有元素的和
print(ret.sum(axis=0))     # 按列求和（每只股票的总收益率）
print(ret.sum(axis=1))     # 按行求和（每一天的总收益率）   

print(ret.max(axis=0))
print(ret.min(axis=1)) 
```

## 4.广播broadcasting

广播：不同形状数组之间，自动做 “虚拟复制扩展”，让它们形状匹配，从而直接运算（+ - * /），不真的复制数据，节省内存。
以下是几个简单例子：

```python
import numpy as np
a = np.array([1,2,3])  # shape (3,)
b = 2                  # 标量，相当于 shape ()
print(a + b)
```

```python
import numpy as np

a = np.array([[1,2,3],
              [4,5,6]]) # shape (2,3)
b = np.array([10,20,30])# shape (3,)
print(a + b)
```

```python
import numpy as np

a = np.array([[1],[2]])   # shape (2,1)
b = np.array([[10,20,30]])# shape (1,3)
print(a + b)
```

下面是广播在量化中的应用：

```python
import numpy as np

ret = np.array([
    [0.01, 0.02, -0.01, 0.00],
    [0.02, -0.01, 0.03, 0.01],
    [-0.01, 0.00, 0.01, 0.02],
])

mu=ret.mean(axis=0)
demeaned_ret = ret - mu
print(demeaned_ret)

sigma = ret.std(axis=0)
standardized_ret = demeaned_ret / sigma
print(standardized_ret)
print(standardized_ret.mean(axis=0))
print(standardized_ret.std(axis=0))

w = np.array([0.1, 0.2, 0.3, 0.4])
portfolio_return = ret @ w
print(portfolio_return)
```

## 5.向量化计算

```python
import time

N = 1_000_000
x = np.random.normal(size=N)

y = np.random.normal(size=N)

# for 循环版
t0 = time.time()
s = 0.0
for i in range(N):
    s += x[i] * y[i]
t1 = time.time()
print("for 循环:", t1 - t0, "结果:", s)

# 向量化版
t0 = time.time()
s2 = np.dot(x, y)
t1 = time.time()
print("向量化:", t1 - t0, "结果:", s2)
```

```python
# 简单收益率 r=P(i+1)/P(i)-1
prices = np.array([100, 101, 102, 101, 103], dtype=float)
rets = prices[1:] / prices[:-1] - 1
print(rets)

# 对数收益率 r=ln(P(i+1)/P(i))
log_rets = np.diff(np.log(prices))
print(log_rets)

# 滚动均值（用pandas里面的rolling方法更方便，详见pandas学习的8.4节）
window = 3
roll_mean=np.array([
    prices[i:i+window].mean() 
    for i in range(len(prices)-window+1)
    ])
print(roll_mean)
```

## 6.矩阵运算

```python
from numpy.linalg import inv,det, eig,svd,solve

A=np.array([[1,2],
            [2,1]])

print("行列式:",det(A))
print("逆矩阵:\n",inv(A))  
print(solve(A,np.array([1,2])))
print("特征值和特征向量:\n",eig(A))
eigvals, eigvecs = eig(A)
print(eigvals)
print(eigvecs)
```

```python
# 收益率矩阵：3 天，4 只股票
ret = np.array([
    [0.01, 0.02, -0.01, 0.00],
    [0.02, -0.01, 0.03, 0.01],
    [-0.01, 0.00, 0.01, 0.02],
])

# 权重向量：4 只股票
w = np.array([0.4, 0.3, 0.2, 0.1])

# 组合每日收益
port_ret = ret @ w
print(port_ret)

# 转置
print(ret.T.shape)   # (4, 3)

# 矩阵乘法：协方差矩阵
# 协方差 Covariance
#协方差衡量两个变量，是不是倾向于 “一起变大、一起变小”，描述两个变量协同波动的方向和幅度。
# ret shape (3, 4)，ret.T shape (4, 3)
# ret.T @ ret shape (4, 4)，即 4 只股票的协方差（未除以 T）
mu=ret.mean(axis=0)
demeaned_ret = ret - mu
cov = demeaned_ret.T @ demeaned_ret / (ret.shape[0]-1)
print(cov)
print(cov.shape)

# 用 np.cov 更标准
cov2 = np.cov(ret, rowvar=False)
print(cov2)
```

```python
# 假设 4 只股票
cov = np.cov(ret, rowvar=False)
ones = np.ones(cov.shape[0])
print(ones)

# 最小方差权重：w = Σ^{-1} 1 / (1^T Σ^{-1} 1)
inv_cov = inv(cov)
w_mv = inv_cov @ ones / (ones @ inv_cov @ ones)
print(w_mv)
print(w_mv.sum())   # 应为 1
```

## 8.for循环与向量化的欧式距离（距离矩阵）

```python
import numpy as np
import time

def euclidean_loop(X,Y):
    n = X.shape[0]
    m = Y.shape[0]
    D = np.zeros((n,m))
    for i in range(n):
        for j in range(m):
            D[i,j] = np.sqrt(np.sum((X[i,:]-Y[j,:])**2))
    return D

def euclidean_vectorized(X,Y):
    x2=np.sum(X**2,axis=1).reshape(-1,1)
    y2=np.sum(Y**2,axis=1).reshape(1,-1)
    xy=X@Y.T
    D=np.sqrt(x2+y2-2*xy)
    return D

rng=np.random.default_rng(42)
X=rng.normal(0,1,size=(200,50))
Y=rng.normal(0,1,size=(100,50))

t0=time.time()
D_loop=euclidean_loop(X,Y)
t1=time.time()
print("Loop time:",t1-t0)

t0=time.time()
D_vectorized=euclidean_vectorized(X,Y)
t1=time.time()
print("Vectorized time:",t1-t0)
```

## 9.组合收益、协方差、风险

```python
rng=np.random.default_rng(42)
T, N = 252, 10

ret = rng.normal(0.005, 0.02, size=(T, N))
w=np.ones(N)/N

port_ret = ret @ w
ann_ret = port_ret.mean() * 252         #年化收益率
ann_vol = port_ret.std() * np.sqrt(252) #年化波动率
sharpe_ratio = ann_ret / ann_vol        #夏普比率
print("Annualized Return:", ann_ret)
print("Annualized Volatility:", ann_vol)    
print("Sharpe Ratio:", sharpe_ratio)

cov = np.cov(ret, rowvar=False)
print("Covariance Matrix:\n", cov)

port_vol = np.sqrt(w @ cov @ w)         #组合波动率
print("Portfolio Volatility:", port_vol)

mrc = cov @ w / port_vol                 #边际风险贡献
print("Marginal Risk Contribution:", mrc)
trc = mrc * w                            #总风险贡献
print("Total Risk Contribution:", trc)
print("Sum of Total Risk Contributions:", trc.sum()) #总风险贡献之和等于组合波动率
rc = trc / port_vol                      #风险贡献占比
print("Risk Contribution:", rc)
print("Sum of Risk Contributions:", rc.sum()) #风险贡献占比之和等于 1
```

## 10.量化中的常见numpy操作

```python
ret = rng.normal(size=(3, 3))
print("Returns:\n", ret)

cum_ret = np.cumprod(1 + ret, axis=0) - 1 # 计算累积收益率
print("Cumulative Returns:\n", cum_ret)

cum_sum = np.cumsum(ret, axis=0)
print("Cumulative Sum:\n", cum_sum)

diff = np.diff(ret, axis=0)
print("Differences:\n", diff)

sorted_ret = np.sort(ret, axis=0)
print("Sorted Returns:\n", sorted_ret)

rank_ret = np.argsort(ret, axis=0)  #argsort返回的是排序后的索引值
print("Ranks:\n", rank_ret)

index_max = np.argmax(ret, axis=0)   #argmax返回的是最大值的索引值
index_min = np.argmin(ret, axis=0)
print("Index of Max Returns:\n", index_max)
print("Index of Min Returns:\n", index_min)

count_positive = np.sum(ret > 0, axis=0)    #统计每个列中正收益的数量
print("Count of Positive Returns:\n", count_positive)

clipped = np.clip(ret, -1, 1)  #将收益率限制在[-0.01, 0.01]之间
print("Clipped Returns:\n", clipped)
```
